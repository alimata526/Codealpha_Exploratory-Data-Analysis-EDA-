## Telecom Customer Churn Analysis

## 1.Introduction

### Business Problem

Customer churn is one of the biggest challenges faced by telecommunication companies. Acquiring new customers is generally more expensive than retaining existing ones. Therefore, understanding why customers leave a service provider is essential for improving customer satisfaction and increasing profitability.
This project analyzes customer data to identify the factors associated with churn. By exploring customer behavior, service usage, and account characteristics, the analysis aims to uncover patterns to inform the company's development of effective customer retention strategies.

## Objectives

The main objectives of this project are to:

-Understand the structure and quality of the telecom customer dataset.
-Perform data cleaning and preprocessing to prepare the data for analysis.
-Explore customer demographics and service usage patterns.
-Identify the factors that influence customer churn.
-Detect trends, patterns, and potential anomalies in the data.
-Test hypotheses using statistical analysis and data visualization.
-Generate actionable business insights and recommendations to reduce customer churn.

## Dataset Description

## Dataset Description

The dataset used in this project is the **Telecom Customer Churn Dataset**, which contains information about customers of a telecommunications company. Each row represents an individual customer and includes details about their account, subscribed services, calling behavior, and whether they have churned (left the company).

The dataset consists of **3,333 customer records** and **20 features**, described below.

| Feature | Description |
|---------|-------------|
| State | Customer's state of residence |
| Account length | Number of days the customer has had the account |
| Area code | Customer's area code |
| International plan | Whether the customer subscribes to an international calling plan |
| Voice mail plan | Whether the customer has a voicemail plan |
| Number vmail messages | Number of voicemail messages |
| Total day minutes | Total daytime call minutes |
| Total day calls | Number of daytime calls |
| Total day charge | Total daytime charges |
| Total eve minutes | Total evening call minutes |
| Total eve calls | Number of evening calls |
| Total eve charge | Total evening charges |
| Total night minutes | Total nighttime call minutes |
| Total night calls | Number of nighttime calls |
| Total night charge | Total nighttime charges |
| Total intl minutes | Total international call minutes |
| Total intl calls | Number of international calls |
| Total intl charge | Total international call charges |
| Customer service calls | Number of customer service calls made |
| Churn | Indicates whether the customer left the company (**True** = churned, **False** = retained) |

### Target Variable

The target variable in this dataset is **Churn**, a binary variable indicating whether a customer has discontinued the company's services.

- **True**: The customer has churned.
- **False**: The customer has remained with the company.

6. Statistical Analysis and Hypothesis Testing

```python
from scipy.stats import chi2_contingency, ttest_ind
```

```python
# Create contingency table
contingency_table = pd.crosstab(df['International_Plan'], df['Churn'])

# Perform Chi-Square Test
chi2, p, dof, expected = chi2_contingency(contingency_table)

print("Chi-Square Statistic:", chi2)
print("P-value:", p)
```

```python
alpha = 0.05

if p < alpha:
    print("Reject the Null Hypothesis")
else:
    print("Fail to Reject the Null Hypothesis")
```

```python
contingency_table = pd.crosstab(df['VoiceMail_Plan'], df['Churn'])

chi2, p, dof, expected = chi2_contingency(contingency_table)

print("P-value:", p)
```

```python
churned = df[df['Churn'] == 1]['Day_Minutes']
not_churned = df[df['Churn'] == 0]['Day_Minutes']

t_stat, p = ttest_ind(churned, not_churned)

print("T-statistic:", t_stat)
print("P-value:", p)
```

```python
churned = df[df['Churn'] == 1]['Customer_Service_Calls']
not_churned = df[df['Churn'] == 0]['Customer_Service_Calls']

t_stat, p = ttest_ind(churned, not_churned)

print("T-statistic:", t_stat)
print("P-value:", p)
```
